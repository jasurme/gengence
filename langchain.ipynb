```python
from langchain.agents import create_agent
from langchain.tools import tool
from langgraph.prebuilt import create_react_agent
from datetime import datetime
from langchain_openai import ChatOpenAI
from langchain_tavily import TavilySearch
from dotenv import load_dotenv
from pydantic import BaseModel
from typing import TypedDict, Union, Annotated, get_type_hints
import operator
from langchain_core.agents import AgentAction,AgentFinish
load_dotenv()

gpt = ChatOpenAI(model='gpt-4o-mini')

@tool
def get_current_date(format: str="%Y-%m-%d %H-%M-%S"):
    "gives current time"
    now = datetime.now()
    return now.strftime(format)

tavily_tool = TavilySearch()

tools = [get_current_date, tavily_tool]

react = create_agent(model=gpt, tools=tools)  

class AgentState(TypedDict):
    input: str
    agent_outcome: Union[AgentAction, AgentFinish, None]
    intermediate_steps: Annotated[list[tuple[AgentAction, str]], operator.add]

```

```python
from pydantic import BaseModel, Field
from typing import Union, Any, Dict
from openai import OpenAI

class ResumeProcessToolCall(BaseModel):
    name: str = Field(description="called tool name")
    input: str = Field(description="input for tool")


class ResumeProcessNormal(BaseModel):
    message: str
    yes_or_no: bool = Field(default=False, description="""ha yoki yo'q savollar uchun. default holatda False; faqat savol ha/yo‘q tarzida javob talab qilsa (masalan, "Yana ... kiritasizmi?", "Boshlashni xohlaysizmi?") shu maydonni True qil. Matnda "ha/yo‘q" varianti yozilmaydi, qaror faqat JSON field orqali beriladi.""")
    file_name: Union[str, None] = Field(description="tooldan olingan file yoki bo‘sh string(None). faqat toolni ishlatganingda buni yoz. boshqa paytlarda None")
    finished: bool

class ResumeProcessOutput(BaseModel):
    output: Union[ResumeProcessToolCall, ResumeProcessNormal]

client = OpenAI()

response = client.responses.parse(
    model="gpt-4o-2024-08-06",
    input=[
        {"role": "system", "content": "answer well"},
        {
            "role": "user",
            "content": "ushbu toollarga access bor. get_time(format: str). user savoli: ahvollaring qanday",
        },
    ],
    text_format=ResumeProcessOutput,
)

event = response.output_parsed.output

if isinstance(event, ResumeProcessNormal):
    print('ResumeProcessNormal: ', event.message)
elif isinstance(event, ResumeProcessToolCall):
    print('ResumeProcessToolCall')
```
